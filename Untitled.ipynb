```python
# Core
import json
import pathlib
import random
import numpy as np

# Imaging
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

# PyTorch
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms




```

```python
# Create a path to the root directory
Root=pathlib.Path("/srv/data/lt2326-h25/a1")
# Create a path to the images directiory
Images= Root / "images"
# Create a path to the annotation file
Annotations = Root / "train.jsonl"
```

```python
# Create path objects for the folders were the processed images and the masks will be saved
out_img=pathlib.Path("images")
out_mask=pathlib.Path("masks")

# Create the folders that will store them
out_img.mkdir(exist_ok=True)
out_mask.mkdir(exist_ok=True)

# List images from the dataset
image_files=sorted([image for image in Images.iterdir()
                   ])
print(len(image_files))
```

```python
# Load all annotations into a dict for easier match with the actual image
# Key : the name of the image
# Value : its annotation
dict_annotations={}

with open (Annotations, "r", encoding="utf8") as annotations:
    for line in annotations:
        obj = json.loads(line)
        dict_annotations[obj["file_name"]]=obj

print("Loaded annotations for", len(dict_annotations), "images.")
```

```python
# Loop through the images and annotations to make sure we only work with the annotations that actually have corresponding image
filtered_dict={}
for img in image_files:
    if img.name in dict_annotations:
        filtered_dict[img.name] = dict_annotations[img.name]

print(len(filtered_dict))
```

```python
NEW_SIZE = (512, 512) # Define the new size, small enough for GPU, big enough for the chars to still show



for fname, annotation in filtered_dict.items():

    # Load image
    image_path = Images / fname
    img = Image.open(image_path).convert("RGB") # Ensure 3 colour channel

    # Create mask 
    # Get width and height from the annotation
    img_w = annotation["width"] 
    img_h = annotation["height"]
    # Create an empty mask, everypixel will be initially background
    mask = Image.new("L", (img_w, img_h), 0)
    draw = ImageDraw.Draw(mask)

    # Loop through each annotation and only draw a white mask if the char is chinese
    # use the annotation info is_chinese
    for sentence in annotation["annotations"]:
        for info in sentence:
            if not info["is_chinese"]:
                continue
            
            xmin, ymin, bw, bh = info["adjusted_bbox"]
            # calculate xmax and ymax since imagedraw.retangle needs to know from which point until which point to draw
            xmax = xmin + bw
            ymax = ymin + bh

            # fill with white the boxes where chinese char are marked 
            draw.rectangle(
                [int(xmin), int(ymin), int(xmax), int(ymax)],
                fill=255
            )

    # Resize
    resized_image = img.resize(NEW_SIZE, resample=Image.Resampling.BILINEAR)
    resized_mask = mask.resize(NEW_SIZE, resample=Image.Resampling.NEAREST)

    # Save the resized image and its mask
    resized_image.save(out_img / fname)
    resized_mask.save(out_mask / fname.replace(".jpg", ".png"))
```

```python


# Paths to processed data
out_img = pathlib.Path("images")

# Ratios for spliting
TRAIN_RATIO = 0.7
VAL_RATIO = 0.15
SEED = 42

# Collect filenames, image+mask
files = [f.name for f in out_img.iterdir() if f.suffix == ".jpg"]

#  Shuffle using seed
random.seed(SEED)
random.shuffle(files)

# Split
n = len(files)
train_end = int(n * TRAIN_RATIO)
val_end = train_end + int(n * VAL_RATIO)

train_files = files[:train_end]
val_files = files[train_end:val_end]
test_files = files[val_end:]

# Save splits to the disc
splits_dir = pathlib.Path("splits")
splits_dir.mkdir(exist_ok=True)

for name, split in [
    ("train.txt", train_files),
    ("val.txt", val_files),
    ("test.txt", test_files),
]:
    with open(splits_dir / name, "w") as f:
        f.write("\n".join(split))
```

```python

class OCRSegmentationDataset(Dataset):
    def __init__(self, split_file, image_dir, mask_dir):
        """
        Args:split_file: path to train.txt / val.txt / test.txt
             image_dir: path to images/
             mask_dir: path to masks/
        """
        # Path where images and masks are saved
        self.image_dir = pathlib.Path(image_dir)
        self.mask_dir = pathlib.Path(mask_dir)

        # Read filenames from split file
        with open(split_file, "r") as f:
            self.files = [line.strip() for line in f]

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        fname = self.files[idx]
        
        # Load image and mask 
        image = Image.open(self.image_dir / fname).convert("RGB")
        mask_path = self.mask_dir / fname.replace(".jpg", ".png")
        mask = Image.open(mask_path).convert("L")
        
        # Convert to Tensors
        image_tensor = transforms.ToTensor()(image)
        mask_tensor = transforms.ToTensor()(mask)
        
        return image_tensor, mask_tensor
```

```python
# Simple CNN model
class CNNModel(nn.Module):
    def __init__(self):
        super(CNNModel, self).__init__()
        
        # Encoder: look from smaller to smaller, and downsize the image from 512x512 to 128x128
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2), 
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)  
        )
        
        # Decoder:  back to 512x512
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(32, 16, kernel_size=2, stride=2), 
            nn.ReLU(),
            nn.ConvTranspose2d(16, 1, kernel_size=2, stride=2),  
            nn.Sigmoid() 
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x
```

```python
# Unet Model
class UNetModel(nn.Module):
    def __init__(self):
        super(UNetModel, self).__init__()

        # Encoder (Downscaling)
        self.enc1 = self.conv_block(3, 16)
        self.enc2 = self.conv_block(16, 32)
        self.pool = nn.MaxPool2d(2, 2)

        # Bottleneck
        self.bottleneck = self.conv_block(32, 64)

        # Decoder (Upscaling)
        self.up2 = nn.ConvTranspose2d(64, 32, kernel_size=2, stride=2)
        self.dec2 = self.conv_block(64, 32) 
        
        self.up1 = nn.ConvTranspose2d(32, 16, kernel_size=2, stride=2)
        self.dec1 = self.conv_block(32, 16)

        self.final = nn.Conv2d(16, 1, kernel_size=1)
        self.sigmoid = nn.Sigmoid()

    def conv_block(self, in_ch, out_ch):
        return nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.ReLU()
        )

    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        
        
        b = self.bottleneck(self.pool(e2))

        #Decoder
        d2 = self.up2(b)
        d2 = torch.cat((d2, e2), dim=1) 
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat((d1, e1), dim=1) 
        d1 = self.dec1(d1)

        return self.sigmoid(self.final(d1))
```

```python
CONFIG = {
    "lr": [0.001, 0.005],
    "epochs": [5, 10],
    "batch_size": 8,
}

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

train_dataset = OCRSegmentationDataset("splits/train.txt", "images", "masks")
val_dataset   = OCRSegmentationDataset("splits/val.txt", "images", "masks")

train_loader = DataLoader(train_dataset, batch_size=CONFIG["batch_size"], shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=CONFIG["batch_size"])

```

```python





models_to_run = [
    ("CNNModel", CNNModel),
    ("UNet", UNetModel)
]


for model_label, model_class in models_to_run:

 
    print(f"Training for model: {model_label}")
    

    best_model_loss = float("inf")   

    for lr in CONFIG["lr"]:
        for num_epochs in CONFIG["epochs"]:

            print(f"\n---Model: {model_label} | LR: {lr} | Epochs: {num_epochs} ---")

            model = model_class().to(device)
            criterion = nn.BCELoss()
            optimizer = torch.optim.Adam(model.parameters(), lr=lr)

            history = []

            #Training
            for epoch in range(num_epochs):

                
                model.train()
                train_loss = 0

                for images, masks in train_loader:
                    images, masks = images.to(device), masks.to(device)

                    outputs = model(images)
                    loss = criterion(outputs, masks)

                    optimizer.zero_grad()
                    loss.backward()
                    optimizer.step()

                    train_loss += loss.item()

                train_loss /= len(train_loader)
                history.append(train_loss)

                #Val
                model.eval()
                val_loss = 0

                with torch.no_grad():
                    for images, masks in val_loader:
                        images, masks = images.to(device), masks.to(device)
                        outputs = model(images)
                        loss = criterion(outputs, masks)
                        val_loss += loss.item()

                val_loss /= len(val_loader)

                print(
                    f"Epoch [{epoch+1}/{num_epochs}] "
                    f"Train Loss: {train_loss:.4f} "
                    f"Val Loss: {val_loss:.4f}"
                )

                # Save bestmodel for each arhitecture
                if val_loss < best_model_loss:
                    best_model_loss = val_loss
                    torch.save(
                        model.state_dict(),
                        f"BEST_{model_label}.pth"
                    )
                    print(f"Best model {model_label} saved!")

            # Plot this run
            plt.plot(
                history,
                label=f"{model_label}-LR{lr}-Ep{num_epochs}"
            )

```

```python

# Evaluation 
def calculate_metrics(preds, masks, threshold):
    # Apply thresholding 
    binary_preds = (preds >= threshold).float()
    
    # Calculate TP, FP, FN for Recall and Precision
    tp = (binary_preds * masks).sum().item()
    fp = (binary_preds * (1 - masks)).sum().item()
    fn = ((1 - binary_preds) * masks).sum().item()
    
    eps = 1e-8
    precision = tp / (tp + fp + eps)
    recall = tp / (tp + fn + eps)
    # IoU 
    iou = tp / (tp + fp + fn + eps)
    
    return {"precision": precision, "recall": recall, "iou": iou}

def evaluate_on_test_set(model, dataloader, thresholds):  
    model.eval()
    results = {}
    for t in thresholds:
        m_list = []
        with torch.no_grad():
            for imgs, masks in dataloader:
                imgs, masks = imgs.to(device), masks.to(device)
                probs = model(imgs)
                m_list.append(calculate_metrics(probs, masks, threshold=t))
        
        # Average results across test set
        avg_precision = np.mean([x["precision"] for x in m_list])
        avg_recall = np.mean([x["recall"] for x in m_list])
        avg_iou = np.mean([x["iou"] for x in m_list])
        results[t] = {"precision": avg_precision, "recall": avg_recall, "iou": avg_iou}
    return results


```

```python
#Test Dataloaders
test_dataset = OCRSegmentationDataset("splits/test.txt", "images", "masks")

test_loader_eval = DataLoader(
    test_dataset,
    batch_size=8,
    shuffle=False
)

test_loader_viz = DataLoader(
    test_dataset,
    batch_size=1,
    shuffle=False
)
```

```python
simple_model = CNNModel().to(device)
simple_model.load_state_dict(
    torch.load("BEST_CNNModel.pth", map_location=device)
)
simple_model.eval()

unet_model = UNetModel().to(device)
unet_model.load_state_dict(
    torch.load("BEST_UNet.pth", map_location=device)
)
unet_model.eval()

```

```python
import numpy as np 
#Evaluate FirstModel
print("CNNModel")
simple_stats = evaluate_on_test_set(simple_model, test_loader_eval, thresholds = [0.05, 0.1, 0.15, 0.2, 0.25]
)

for t, metrics in simple_stats.items():
    print(f"Threshold {t}: Precision: {metrics['precision']:.3f}, Recall: {metrics['recall']:.3f}, IoU: {metrics['iou']:.3f}")

#Evaluate SecondModel
print("UNetModel")
unet_stats = evaluate_on_test_set(unet_model, test_loader_eval, thresholds = [0.05, 0.1, 0.15, 0.2, 0.25]
)

for t, metrics in unet_stats.items():
    print(f"Threshold {t}: Precision: {metrics['precision']:.3f}, Recall: {metrics['recall']:.3f}, IoU: {metrics['iou']:.3f}")
```

```python
def visualize_model_comparison(
    image_tensor,
    mask_tensor,
    simple_model,
    unet_model,
    threshold_simple=0.05,
    threshold_unet=0.05
):
    simple_model.eval()
    unet_model.eval()

    with torch.no_grad():
        simple_prob = simple_model(image_tensor.unsqueeze(0).to(device))[0, 0].cpu().numpy()
        unet_prob   = unet_model(image_tensor.unsqueeze(0).to(device))[0, 0].cpu().numpy()

    img_np  = image_tensor.permute(1, 2, 0).numpy()
    gt_mask = mask_tensor.squeeze().numpy()

    simple_pred = (simple_prob >= threshold_simple).astype(np.float32)
    unet_pred   = (unet_prob >= threshold_unet).astype(np.float32)

    
    fig, axes = plt.subplots(1, 5, figsize=(22, 5))

    axes[0].imshow(img_np)
    axes[0].set_title("Original Image")
    axes[0].axis("off")

    axes[1].imshow(gt_mask, cmap="gray")
    axes[1].set_title("Ground Truth Mask")
    axes[1].axis("off")

    axes[2].imshow(img_np)
    axes[2].imshow(simple_prob, cmap="hot", alpha=0.5)
    axes[2].set_title("CNNModel")
    axes[2].axis("off")

    axes[3].imshow(img_np)
    axes[3].imshow(unet_prob, cmap="hot", alpha=0.5)
    axes[3].set_title("UNet")
    axes[3].axis("off")

    # Overlay agreement
    overlay = simple_pred + 2 * unet_pred
    axes[4].imshow(overlay, cmap="viridis")
    axes[4].set_title("Thresholded Agreement\n(CNN | UNet)")
    axes[4].axis("off")

    plt.tight_layout()
    plt.show()
```

```python
images, masks = next(iter(test_loader_viz))
sample_img = images[0]
sample_mask = masks[0]

visualize_model_comparison(
    sample_img,
    sample_mask,
    simple_model,
    unet_model,
    threshold_simple=0.05,
    threshold_unet=0.05
)
```
